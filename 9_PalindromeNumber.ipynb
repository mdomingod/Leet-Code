## Built-in Types: str

````python
# str()
s = str(x)
s[::-1]  # Reverse the string

class Solution:
	def isPalindrome(self, x: int) -> bool:
		s = str(x)
		s[::-1]  # Reverse the string
		if s == s[::-1]:
			return True
		return False
````
