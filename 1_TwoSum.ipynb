```python
# SOLUTION !!!!

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        num_map= {}

        for i, num in enumerate(nums): # i --> index, nums --> número
            complement = target - num 

            if not complement in num_map: # si el dict està buuit: 
         
                num_map[num] = i  # Afegeix una key: value
    
            else: # si no està buit
                if complement in num_map: # i el complement es troba en el dict:
                    return[num_map[complement], i]
```
